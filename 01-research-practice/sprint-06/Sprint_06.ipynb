## A. Problem Situation — чи достатньо отримати графіки для наукової статті?

У Sprint 5 ми побудували криві якості й деградації та побачили варіативність за різних реалізацій шуму. Чи можна тепер написати: «один алгоритм стійкіший за інший»? Але що саме означає це твердження — для обраного датасету та його розбиття, за вибраного механізму шуму, чи взагалі для всіх задач класифікації?

Перш ніж переходити до рукопису, з'ясуємо:

- чи можна повторно отримати **ті самі числові результати** зі збережених моделей і конфігурації;
- які твердження безпосередньо випливають із таблиць, а які є інтерпретацією;
- чи відповідаємо на **кожне** RQ, а не лише описуємо найвиразніший графік;
- які чинники обмежують висновки, навіть якщо програмних помилок немає;
- що можемо зіставити з літературою Sprint 2, не приписуючи публікаціям неперевірених висновків.

**Проблема спринту:** науковий результат — це не графік і не автоматично створений висновок, а **відтворюваний доказовий зв'язок між Research Question, даними, інтерпретацією та межами твердження**.

## B1. Essential Theory — Results, Discussion та Limitations виконують різні функції

**Results** — *що отримано*: значення показників, характер залежностей, варіативність, посилання на таблиці й рисунки. Формулюємо фактичні твердження без причинних пояснень, яких експеримент не перевіряв.

**Discussion** — *що це означає для RQ і H1*: зіставляємо спостереження з припущенням, пояснюємо можливі причини лише як **гіпотези для подальшої перевірки**, пов'язуємо висновки з перевіреними джерелами Literature Review.

**Limitations** — *чого наш результат не доводить*: фіксоване розбиття, один датасет, конкретні конфігурації моделей, один тип шуму, обмежене число повторень, обрані метрики.

| Частина | Коректний тип твердження | Чого уникаємо |
|---|---|---|
| Results | «За $\sigma=0.6$ середня деградація моделі X становила …» | «X від природи нестійкий» |
| Discussion | «Спостережені профілі різняться за досліджених умов» | «Доведено універсальну перевагу X» |
| Limitations | «Повторення шуму на фіксованому Test Set оцінюють варіативність шуму» | «Результат обов'язково відтвориться всюди» |

**Правило переходу:** *числове спостереження → порівняння за RQ → інтерпретація → обмеження*. У тексті статті посилання на рисунок або таблицю має допомагати читачеві перевірити твердження.

## B2. Essential Theory — що означає перевірити відтворюваність результатів?

Ми розрізняємо:

1. **Контроль цілісності артефактів:** перевіряємо, що є потрібні CSV, конфігурація, моделі, seed та індекси Test Set; збережені моделі не змінилися, а таблиці не суперечать одна одній.
2. **Контрольне повторення обчислень:** використовуючи *збережені* моделі та *той самий* Test Set, повторно обчислюємо `Q` і `D` для всіх умов **без tuning і перенавчання**; зіставляємо з `measurements.csv`.
3. **Документування середовища:** фіксуємо версії Python, NumPy, Pandas, scikit-learn, спосіб запуску та виявлені відхилення.

Перші два кроки перевіряють **обчислювальну відтворюваність у зафіксованих умовах**. Це перевірка повторюваності розрахунків за збереженими артефактами.

Хеш SHA-256 допомагає перевірити, **чи змінювалися байти файла**, але сам по собі не підтверджує наукову коректність моделі. Для чисел із CSV порівнюємо значення з малою обґрунтованою похибкою, а не вимагаємо буквального збігу текстового запису десяткових дробів.

**Важливо:** повторний розрахунок уже виміряної умови для аудиту — це не нова незалежна реалізація шуму і не додатковий рядок Experimental Matrix.

## B3. Essential Theory — від чисел до відповідей на RQ

У нашому прикладі зберігаємо структуру, погоджену у Sprint 3:

- **RQ1:** показуємо, як змінюється `Q(m,σ)`; описуємо всю криву, а не лише крайні точки.
- **RQ2:** показуємо профілі `D(m,σ)=Q(m,0)-Q(m,σ)` для трьох алгоритмів; зіставляємо **спостережені** відмінності без окремого парного статистичного аналізу.
- **RQ3:** показуємо розкид за п'ятьма реалізаціями тестового шуму; визначаємо, наскільки результат залежить від seed **за фіксованого Test Set**.

Для кожного RQ з'єднуємо три елементи: **питання → конкретна таблиця/графік → обмежена відповідь**.

**Оцінка H1:** якщо різні профілі деградації спостерігаються за наших умов, можемо сказати, що це *узгоджується з H1 на описовому рівні*. Без попередньо встановленого критерію й коректного інферентного методу неможливо оголошувати формальну статистичну значущість; подібність профілів не доводить їхньої еквівалентності.

## B4. Essential Theory — аналіз обмежень, а не перелік застережень

Обмеження випливають із **рішень Experimental Protocol**. Розглянемо наші рішення:

| Рішення в нашому дизайні | Що воно забезпечує | Яке обмеження залишається |
|---|---|---|
| Один фіксований Training/Test split | Контрольоване порівняння | Не оцінено залежність від інших вибірок |
| Оптимізація гіперпараметрів за чистою macro-F1 | Обґрунтовані фіксовані моделі | Не досліджено оптимізацію саме стійкості |
| Гаусівський шум у стандартизованих тестових ознаках | Чіткий керований вплив | Не досліджено інші типи шуму або спотворення навчальних даних |
| П'ять seed на тому самому Test Set | Оцінка варіативності випадкового шуму | Це не п'ять незалежних датасетів |
| `load_wine` та три алгоритми | Доступний відтворюваний приклад | Обмежено перенесення на інші задачі й алгоритми |

**Щоб визначити межу висновку,** для кожної ключової фрази запитаємо, *чи змінювали цей чинник у нашому експерименті?* Якщо ні, не можемо без окремих даних стверджувати, що ефект збережеться за його зміни.

Опис обмежень не знецінює дослідження. Він відділяє те, що ми **встановили**, від того, що **можна перевірити** в наступних експериментах.

## B5. Essential Theory — літературний контекст і наукове формулювання

У Sprint 2 ми сформували Literature Review Matrix та обґрунтували Research Gap. У *Discussion* повертаємося до **реально перевірених** публікацій, позначені в навчальному прикладі умовними P1–P4.

Зв'язок із літературою вибудовуємо за аспектами:

1. **Зіставність умов:** чи однакові тип шуму, етап його внесення, датасет, алгоритми та метрики?
2. **Узгодженість результатів:** чи підтримують опубліковані спостереження нашу інтерпретацію, чи виявляють відмінності?
3. **Пояснення розбіжностей:** чи достатньо даних, щоб пояснити їх, чи це лише можливі причини?
4. **Внесок:** яке *конкретне* знання додає наш результат **у межах дослідженого сценарію**?

Якщо фактичні джерела для зіставлення ще не внесені до власної бібліотеки, **залишаємо місце для посилань** і реалізуємо розширений пошук наукових джерел за означеними критеріями.

Наприкінці отримуємо чотири частини майбутнього наукового тексту: **Results → Discussion → Limitations → Reproducibility Statement**. Їх поєднаємо з іншими розділами статті у Sprint 7.

## C1. Guided Research Example — читаємо результати Sprint 5 як незмінне джерело

Починаємо не з повторного експерименту, а зі **збережених вимірювань**. Каталог `research_artifacts/sprint5/` містить `measurements.csv`, `summary.csv`, `results_manifest.json` і графіки.

Перевіряємо вхідні артефакти і створимо **окремий** каталог `research_artifacts/sprint6/` для аудиту. Він не повинен перезаписувати результати попередніх спринтів.

```python
from pathlib import Path
import json
import hashlib
import platform
import numpy as np
import pandas as pd
import sklearn
import joblib
from sklearn.datasets import load_wine
from sklearn.metrics import f1_score
from IPython.display import display

ARTIFACT_DIR = Path("research_artifacts")
RESULT_DIR = ARTIFACT_DIR / "sprint5"
REVIEW_DIR = ARTIFACT_DIR / "sprint6"

# Перевіряємо вхідні артефакти, не створюючи нових вимірювань.
required = [RESULT_DIR / "measurements.csv", RESULT_DIR / "summary.csv",
            RESULT_DIR / "results_manifest.json", ARTIFACT_DIR / "manifest.json",
            ARTIFACT_DIR / "split_indices.npz"]
required += [ARTIFACT_DIR / f"model_{name}.joblib" for name in ("LR", "RF", "GB")]
missing = [str(p) for p in required if not p.is_file()]
if missing:
    raise FileNotFoundError(f"Бракує результатів Sprint 4–5: {missing}")

manifest4 = json.loads((ARTIFACT_DIR / "manifest.json").read_text(encoding="utf-8"))
manifest5 = json.loads((RESULT_DIR / "results_manifest.json").read_text(encoding="utf-8"))
config = manifest4["config"]
measurements = pd.read_csv(RESULT_DIR / "measurements.csv")
summary_saved = pd.read_csv(RESULT_DIR / "summary.csv")
REVIEW_DIR.mkdir(parents=True, exist_ok=True)
print("Завантажено:", len(measurements), "рядки вимірювань і", len(summary_saved), "рядків зведення")
```

## C2. Guided Research Example — аудит цілісності результатів

На цьому етапі перевіряємо повноту, цілісність і відповідність збережених результатів експериментальному протоколу. Повторні обчислення метрик не виконуємо — використовуємо результати, отримані у Sprint 5.

```python
# Перевіряємо матрицю без дублювання пілоту та формулу деградації.
assert len(measurements) == 63
assert measurements.groupby(["model", "sigma"]).ngroups == 15
assert not measurements.duplicated(["model", "sigma", "seed"]).any()
assert set(measurements.model) == {"LR", "RF", "GB"}
assert set(np.round(measurements.sigma.unique(), 8)) == set(config["noise_levels"])
assert (measurements.Q_noisy.between(0, 1) & measurements.Q_clean.between(0, 1)).all()
assert np.allclose(measurements.D, measurements.Q_clean - measurements.Q_noisy, atol=1e-10)
assert (measurements.source == "pilot_sprint4").sum() == 3
for model in ("LR", "RF", "GB"):
    subset = measurements[measurements.model == model]
    assert len(subset[subset.sigma == 0]) == 1
    for sigma in config["noise_levels"][1:]:
        group = subset[np.isclose(subset.sigma, sigma)]
        assert sorted(group.seed.dropna().astype(int).tolist()) == sorted(config["noise_seeds"])

# Перевіряємо перерахунок summary з measurements.csv.
recomputed_summary = (measurements.groupby(["model", "sigma"], as_index=False)
                      .agg(n=("Q_noisy", "size"), Q_mean=("Q_noisy", "mean"),
                           Q_sd=("Q_noisy", "std"), D_mean=("D", "mean"),
                           D_sd=("D", "std")))
merged = recomputed_summary.merge(summary_saved, on=["model", "sigma"],
                                  suffixes=("_check", "_saved"), validate="one_to_one")
assert len(merged) == 15
for field in ("n", "Q_mean", "Q_sd", "D_mean", "D_sd"):
    assert np.allclose(merged[f"{field}_check"], merged[f"{field}_saved"],
                       equal_nan=True, atol=1e-10, rtol=0), field

def sha256_file(path):
    h = hashlib.sha256()
    with path.open("rb") as stream:
        for part in iter(lambda: stream.read(1024 * 1024), b""):
            h.update(part)
    return h.hexdigest()

for name in ("LR", "RF", "GB"):
    assert sha256_file(ARTIFACT_DIR / f"model_{name}.joblib") == manifest5["model_sha256"][name]
print("PASS: структура вимірювань, summary і хеші моделей узгоджені.")
```

## C3. Guided Research Example — контрольне повторення без перенавчання

Переходимо від перевірки файлів до **повторного обчислення**: завантажимо саме ті моделі та Test Set, які використовувалися в Sprint 5, відновимо кожну реалізацію шуму за `σ` та `seed` і заново обчислимо `Q` та `D`.

**Порядок контролю:** *збережена модель → зафіксований Test Set → стандартизація → зафіксований шум → прогноз → macro-F1 → звірка з CSV*.

Не викликаємо `fit()` або `GridSearchCV`. Кожне повторне оцінювання — лише **аудит уже виконаного експерименту**. Файли `measurements.csv` та `summary.csv` не переписуємо.

```python
# Відновлюємо Test Set і збережені навчені моделі.
raw = load_wine(as_frame=True)
assert len(raw.data) == manifest4["dataset_rows"]
assert raw.data.columns.tolist() == manifest4["feature_names"]
with np.load(ARTIFACT_DIR / "split_indices.npz") as indexes:
    train_idx, test_idx = indexes["train"], indexes["test"]
assert set(train_idx).isdisjoint(set(test_idx))
X_test = raw.data.iloc[test_idx].copy()
y_test = raw.target.iloc[test_idx].copy()
models = {name: joblib.load(ARTIFACT_DIR / f"model_{name}.joblib")
          for name in ("LR", "RF", "GB")}
reference_scaler = models["LR"].named_steps["scale"]
for pipeline in models.values():
    assert np.allclose(pipeline.named_steps["scale"].mean_, reference_scaler.mean_)
    assert np.allclose(pipeline.named_steps["scale"].scale_, reference_scaler.scale_)
z_clean = reference_scaler.transform(X_test)

# Генератор точно відтворює зафіксований механізм тестового шуму.
def noise_on_standardized(z, sigma, seed):
    rng = np.random.default_rng(int(seed))
    return np.asarray(z, dtype=float) + rng.normal(0.0, float(sigma), size=z.shape)

# Повторюємо усі 63 оцінювання; жодна модель не перенавчається.
check_rows = []
for sigma in config["noise_levels"]:
    seeds = [None] if sigma == 0 else config["noise_seeds"]
    for seed in seeds:
        z = z_clean if seed is None else noise_on_standardized(z_clean, sigma, seed)
        for name, pipeline in models.items():
            prediction = pipeline.named_steps["model"].predict(z)
            q = f1_score(y_test, prediction, average="macro")
            check_rows.append({"model": name, "sigma": float(sigma),
                               "seed": seed, "Q_noisy_check": float(q)})
checked = pd.DataFrame(check_rows)
compared = measurements.merge(checked, on=["model", "sigma", "seed"],
                              how="outer", validate="one_to_one", indicator=True)
assert len(compared) == 63 and (compared._merge == "both").all()
compared["absolute_difference"] = (compared.Q_noisy - compared.Q_noisy_check).abs()
max_difference = float(compared.absolute_difference.max())
passed = bool(np.allclose(compared.Q_noisy, compared.Q_noisy_check,
                         atol=1e-10, rtol=0))
assert passed, f"Результати не відтворилися: максимальне відхилення {max_difference}"
print(f"PASS: контрольне повторення 63 оцінок; найбільше |ΔQ| = {max_difference:.3g}")
```

## C4. Guided Research Example — фіксуємо результат перевірки відтворюваності

Створимо `Reproducibility_Report_v1.md` із **фактичним результатом контрольного запуску**: кількість звірених оцінок, максимальне відхилення, відповідність хешів, версії середовища. Звіт не замінює експертної оцінки гіпотези.

**Позитивний результат аудиту означає,** що вдалося відтворити обчислення в межах конкретних даних, моделей, seed та середовища.

## C5. Guided Research Example — готуємо фактичні свідчення для Results

```python
# Виводимо фактичні значення Results.
max_sigma = max(config["noise_levels"])
clean = summary_saved[np.isclose(summary_saved.sigma, 0)][["model", "Q_mean"]]
end = summary_saved[np.isclose(summary_saved.sigma, max_sigma)][
    ["model", "Q_mean", "Q_sd", "D_mean", "D_sd"]]
results_table = clean.merge(end, on="model", suffixes=("_clean", "_end"))
results_table = results_table.rename(columns={
    "Q_mean_clean": "Q(m,0)", "Q_mean_end": "Q_mean(sigma_max)",
    "Q_sd": "Q_sd(sigma_max)", "D_mean": "D_mean(sigma_max)",
    "D_sd": "D_sd(sigma_max)"})
display(results_table.round(4))

# Перевіряємо поведінку профілю на всіх, а не лише крайніх рівнях.
for name in ("LR", "RF", "GB"):
    s = summary_saved[summary_saved.model == name].sort_values("sigma")
    monotone = np.all(np.diff(s.Q_mean.to_numpy()) <= 1e-12)
    print(f"{name}: Q за рівнями sigma = {s.Q_mean.round(4).tolist()}; "
          f"середня Q невисхідна: {monotone}")
```

## C7. Guided Research Example — Discussion: інтерпретація RQ та оцінювання H1

У C6 ми представили результати експерименту: абсолютну якість класифікації, її деградацію зі збільшенням інтенсивності шуму та варіативність між повторними вимірюваннями.

Тепер розглянемо, **які відповіді на Research Questions дають отримані результати та що вони означають для нашої гіпотези H1**.

### RQ1. Як змінюється якість класифікації зі збільшенням шуму?

За результатами експерименту усі три моделі демонструють зниження середньої якості класифікації зі збільшенням інтенсивності гаусівського шуму в тестових ознаках.

Logistic Regression і Random Forest початково мають однакову macro-F1 = 1.0000, тоді як Gradient Boosting — 0.9453.

За найбільшого дослідженого рівня шуму $\sigma=0.6$ середня якість знижується до:

- **Logistic Regression:** 0.9661;
- **Random Forest:** 0.9235;
- **Gradient Boosting:** 0.8226.

Графік `Q(m,σ)` показує, що середні значення якості для всіх трьох алгоритмів послідовно не зростають зі збільшенням шуму. Отже, вплив шуму спостерігається не лише за найбільшої його інтенсивності, а протягом дослідженого діапазону.

**Відповідь на RQ1:** зі збільшенням інтенсивності шуму середня якість класифікації усіх трьох моделей знижується, але величина погіршення залежить від алгоритму.

### RQ2. Чи відрізняються профілі деградації між алгоритмами?

Для порівняння стійкості використовуємо показник $D(m,\sigma)$, який характеризує зменшення якості кожної моделі відносно її власного базового рівня.

За $\sigma=0.6$ отримано:

| Алгоритм | Середня деградація $D$ |
|---|---:|
| Logistic Regression | 0.0339 |
| Random Forest | 0.0765 |
| Gradient Boosting | 0.1227 |

Значення показують, що за однакової інтенсивності шуму моделі демонструють різне абсолютне зниження macro-F1 відносно власного початкового рівня.

Найменшу середню деградацію демонструє Logistic Regression, найбільшу — Gradient Boosting. Random Forest займає проміжне положення.

Графік `D(m,σ)` дозволяє простежити ці відмінності протягом дослідженого діапазону шуму. Таким чином, відмінності між алгоритмами характеризуються не лише кінцевими значеннями якості, а й величиною її зменшення відносно початкового рівня.

**Відповідь на RQ2:** у проведеному експерименті спостерігаються відмінності профілів деградації трьох алгоритмів. Logistic Regression демонструє меншу середню деградацію за найбільшого рівня шуму, ніж Random Forest і Gradient Boosting.

### RQ3. Наскільки результати залежать від випадкової реалізації шуму?

Для кожного додатного рівня шуму виконано п'ять повторних вимірювань із різними seed.

За $\sigma=0.6$ стандартні відхилення показника деградації становлять:

| Алгоритм | Стандартне відхилення $D$ |
|---|---:|
| Logistic Regression | 0.0232 |
| Random Forest | 0.0314 |
| Gradient Boosting | 0.0307 |

Отже, навіть за фіксованої інтенсивності шуму оцінка деградації змінюється залежно від конкретної реалізації випадкових спотворень.

Random Forest і Gradient Boosting демонструють дещо більший абсолютний розкид деградації, ніж Logistic Regression, за найбільшого дослідженого рівня шуму.

Це означає, що окреме вимірювання не повністю характеризує поведінку моделі. Середні значення дозволяють описати спостережувану тенденцію, а стандартні відхилення — оцінити її варіативність.

**Відповідь на RQ3:** результати залежать від конкретної реалізації шуму; величина цієї варіативності також відрізняється між алгоритмами.

### Оцінювання гіпотези H1

Нагадаємо формулювання H1:

> За однакових експериментальних умов Logistic Regression, Random Forest і Gradient Boosting демонструватимуть різні профілі деградації якості зі збільшенням інтенсивності шуму в тестових числових ознаках.

Отримані результати показують:

1. Якість усіх трьох моделей зменшується зі збільшенням інтенсивності шуму.
2. Величина деградації відрізняється між алгоритмами.
3. Повторні вимірювання виявляють варіативність, яку необхідно враховувати під час порівняння результатів.

**Таким чином, отримані результати узгоджуються з H1 на рівні описового аналізу:** за досліджених умов три алгоритми демонструють різну чутливість до контрольованого шуму тестових ознак.

Водночас наш експериментальний протокол не передбачав окремої процедури статистичної перевірки міжмодельних відмінностей. Тому наведений висновок характеризує спостережувані закономірності, але не є твердженням про формально встановлену статистичну значущість цих відмінностей.

### Підсумок Discussion

Ми отримали не просто три набори значень macro-F1, а **експериментальні свідчення про відмінності поведінки алгоритмів за контрольованого спотворення вхідних даних**.

Результати дозволяють описати характер зменшення якості, порівняти профілі деградації та оцінити їхню варіативність. Саме ці спостереження становлять основу нашої відповіді на H1.

## G. Solutions to Training Exercises — відповіді для самоперевірки

### Вправа 1

Можливе формулювання: «За дослідженої інтенсивності шуму на цьому Test Set середня деградація macro-F1 моделі A була меншою, ніж у B». Воно не дозволяє стверджувати універсальну перевагу, оскільки інші датасети, розбиття, моделі та шуми не досліджувалися.

### Вправа 2

Абсолютне відхилення — приблизно `1 × 10⁻¹¹`, тобто менше за допуск `1 × 10⁻¹⁰`; за цим критерієм оцінки узгоджені. Отже, за встановленим допуском контрольне повторення є успішним.

### Вправа 3

Приклад: «За визначених умов спостережені профілі деградації трьох алгоритмів відрізняються, що на описовому рівні узгоджується з H1». «Висновок стосується лише обраного датасету, розбиття та механізму тестового шуму; п'ять повторень не оцінюють узагальнюваність на інші вибірки».
