# Sprint 1 — Research Problem & Hypothesis
## Від обраної теми до перевірюваної гіпотези та дослідницьких питань

**AI Research Practices · Науково-дослідна практика · Спринт 1 із 7**

> **Стартова умова:** тему дослідження вже обрано студентом і погоджено з науковим керівником. Вибір теми не є завданням цього спринту.

**Мета:** навчитися уточнювати проблему в межах обраної теми, формулювати дослідницьку гіпотезу та перетворювати її на систему дослідницьких питань, з яких випливають цілі й попередня логіка перевірки.

**Результат:** `Research Problem Statement` — короткий документ (орієнтовно 1–2 сторінки) з темою, проблемою, гіпотезою, дослідницькими питаннями, цілями, попереднім підходом до перевірки й межами дослідження.

**Важливо:** цей спринт не вимагає програмування, отримання експериментальних результатів або підтвердження гіпотези.

## Маршрут спринту

1. Від теми — до конкретної проблеми.
2. Від проблеми — до гіпотези.
3. **Від гіпотези — до дослідницьких питань (центральна частина).**
4. Від питань — до дослідницьких цілей та попередньої перевірки.
5. Самостійна постановка власного дослідження.
6. Самоперевірка і підготовка результату.

**Метод роботи:** пояснення → наскрізний приклад → коротке тренувальне завдання → самостійна робота. \
Навчальний приклад демонструє логіку міркування, але не підмінює власний зміст дослідження студента.

## A. Проблемна ситуація

**Обрана тема наскрізного прикладу:** *Стійкість алгоритмів машинного навчання до шуму у вхідних даних*.

Уявімо систему класифікації за показниками датчиків. У лабораторних умовах дані можуть бути чистими, а під час реальної експлуатації вимірювання спотворюються. Виникає ризик, що класифікатор змінюватиме свої рішення через шум.

**Запитання для роздуму:**
- Що саме означає «стійкість» алгоритму в цій ситуації?
- Чи всі алгоритми однаково реагують на спотворення?
- Які спостереження дозволили б обґрунтовано порівняти їхню поведінку?

**Примітка про наукову доброчесність.** Ситуація навчальна; жодних виміряних результатів тут не наведено. Будь-які твердження про поведінку алгоритмів надалі треба перевіряти на реальних експериментальних даних.

## B. Мінімальна теорія: що ми формулюємо

| Поняття | Робоче значення у спринті | Типова помилка |
|---|---|---|
| **Research Topic** | Загальна предметна область вже погодженого дослідження | Вважати назву технології готовою проблемою |
| **Research Problem** | Конкретна невизначеність, обмеження знань або практична суперечність, що потребує вивчення | Писати лише «підвищити точність» без уточнень |
| **Research Hypothesis** | Припущення про спостережувану закономірність або зв'язок, яке можна перевірити й потенційно не підтвердити | Формулювати гасло «метод є найкращим» |
| **Research Question (RQ)** | Питання, відповідь на яке дає потрібне свідчення для оцінювання гіпотези | Підмінювати питання інструкцією «навчити модель» |
| **Research Objective** | Дія, яку треба виконати для отримання відповіді на RQ | Плутати мету з передбаченим результатом |
| **Verification Approach** | Загальний задум порівняння, вимірювання та контролю умов | Передчасно оголошувати результат без експерименту |

**Дві важливі відмінності:**

- **Гіпотеза** відповідає на запитання: *яку залежність ми припускаємо?*
- **Дослідницькі питання** відповідають на запитання: *що треба встановити, щоб оцінити це припущення?*

У реальній роботі питання й гіпотези уточнюються ітеративно. Для навчання ми свідомо опрацьовуємо напрям **гіпотеза → RQ → цілі → перевірка**.

## C1. Наскрізний приклад: від теми до проблеми

**Тема:** «Стійкість алгоритмів машинного навчання до шуму у вхідних даних».

**Недостатнє формулювання:** «Дослідити алгоритми машинного навчання».

Чому недостатнє? Не визначено ні властивість, яку досліджуватимемо, ні умови, ні порівнювані об'єкти.

**Робоче формулювання проблеми:**

> У вибраному сценарії класифікації невідомо, як інтенсивність шуму в числових вхідних ознаках впливає на якість різних алгоритмів та чи відрізняється їхня чутливість за однакових умов оцінювання.

Це **попередня постановка**, а не твердження про відсутність відповідних досліджень у літературі. На Sprint 2 її уточнимо за результатами огляду публікацій.

**Звуження масштабу:** задача класифікації, три вибрані алгоритми (Logistic Regression, Random Forest, Gradient Boosting), числові ознаки, контрольований тип шуму, узгоджені правила оцінювання.

## C2. Наскрізний приклад: формулюємо гіпотезу

**Робоча гіпотеза H1:**

> За однакового експериментального протоколу Logistic Regression, Random Forest і Gradient Boosting демонструватимуть різну чутливість якості класифікації до зростання рівня шуму в числових вхідних ознаках.

**Що тут перевірюване?**
- **Об'єкти порівняння:** три алгоритми.
- **Змінюваний чинник:** інтенсивність шуму.
- **Спостережуваний ефект:** зміна якості класифікації.
- **Умова порівнянності:** однаковий протокол оцінювання.
- **Можливий результат, що не підтримує H1:** відмінності чутливості не виявляються в межах визначеної точності оцінювання.

Гіпотеза не стверджує наперед, який алгоритм «переможе». Її може підтримати або не підтримати експеримент. Обрані моделі і вид шуму в цій вправі — елементи дизайну прикладу, а не доведені універсальні рекомендації.

## C3. Центральний метод: як із H1 отримати дослідницькі питання

На цьому етапі можлива типова помилка: просте переформулювання гіпотези у вигляді питання.

**Невдалий варіант:** «Чи демонструють алгоритми різну чутливість до шуму?»

Це формально допустиме загальне питання, але саме по собі воно **не підказує**, які спостереження і порівняння треба виконати.

**Краще — розкласти гіпотезу на інформаційні потреби.**

Для цього послідовно запитуємо:

1. **Що змінюємо?** Рівень шуму у вхідних ознаках.
2. **Що вимірюємо?** Якість класифікації та її зміну відносно якості класифікації без штучного шуму.
3. **Що порівнюємо?** Реакцію різних алгоритмів на той самий чинник.
4. **Наскільки надійний висновок?** Чи відтворюються спостережувані відмінності за повторних оцінювань?
5. **Що може спростувати припущення?** Відсутність достатніх свідчень відмінності чутливості у визначених умовах.

Цей процес перетворює одну загальну гіпотезу на кілька **операційно корисних питань**.

## C4. Від гіпотези до RQ: покрокова декомпозиція

| Фрагмент гіпотези / потреба | Дослідницьке питання | Навіщо потрібна відповідь |
|---|---|---|
| «...до зростання рівня шуму» | **RQ1.** Як змінюється якість класифікації кожного алгоритму зі збільшенням інтенсивності шуму? | Встановити реакцію кожної моделі, перш ніж порівнювати їх між собою |
| «...різну чутливість» | **RQ2.** Чи відрізняються алгоритми за величиною деградації якості відносно умов без доданого шуму? | Безпосередньо перевірити ключове твердження H1 |
| «...за однакового протоколу» | **RQ3.** Наскільки стабільними є виявлені відмінності за повторних оцінювань з різними випадковими реалізаціями шуму? | Не сплутати випадкове коливання з відтворюваною відмінністю |
| Деталізація залежності | **RQ4 (необов'язкове).** У яких діапазонах інтенсивності шуму відмінності між алгоритмами найбільші? | Уточнити характер встановленої залежності, якщо ресурсів достатньо |

**RQ2 — центральне питання:** саме воно безпосередньо пов'язане з H1. RQ1 і RQ3 забезпечують підстави для відповіді; RQ4 розширює опис, але не обов'язкове для першого невеликого дослідження.

**Не треба формулювати якомога більше RQ.** Для короткої статті достатньо 2–3 пов'язаних питань, якщо вони разом дозволяють оцінити гіпотезу.

## C5. Перевірка якості RQ: п'ять тестів

Для **кожного** сформульованого питання перевіряємо:

1. **Зв'язок із H1.** Яку саме частину гіпотези допоможе оцінити відповідь?
2. **Спостережуваність.** Які дані або порівняння потрібні для відповіді?
3. **Однозначність.** Чи зрозуміло, які об'єкти, чинники й результати маються на увазі?
4. **Відкритість результату.** Чи можлива відповідь, яка не підтвердить припущення?
5. **Реалістичність.** Чи можна відповісти в межах одного невеликого дослідження?

**Антиприклади:**

| Слабке формулювання | Чому слабке | Як покращити |
|---|---|---|
| «Чи хороший Random Forest?» | «Хороший» не має визначеного критерію | «Як змінюється F1-score Random Forest за зростання рівня шуму?» |
| «Як навчити три моделі?» | Це технологічне завдання, а не дослідницьке питання | «Чи відрізняється деградація якості трьох моделей за однакового протоколу?» |
| «Чому Random Forest завжди найкращий?» | Містить недоведене припущення про результат | «Який алгоритм демонструє менше зниження якості у визначених умовах?» |
| «Як працює машинне навчання?» | Занадто широке для експериментальної відповіді | Звузити до вимірюваної властивості певних моделей |

**Правило:** дослідницьке питання має вимагати отримання нового для нашої роботи свідчення, а не переказу визначення з підручника.

## Тренувальна вправа 1 — Розрізняємо типи формулювань

Для кожного твердження визначаємо, чи це **гіпотеза**, **дослідницьке питання**, **дослідницька дія** або **передчасний висновок**. Обґрунтовуємо відповідь.

A. «Побудувати графік залежності F1-score від рівня шуму».

B. «Відмінності чутливості алгоритмів до шуму є відтворюваними».

C. «Чи зберігаються відмінності деградації якості за повторних експериментів?»

D. «Random Forest підтверджено як найстійкіший алгоритм» (експерименти ще не виконані).

Відповідь для самоперевірки наведено в розділі H після контрольних питань.

## Тренувальна вправа 2 — Визначаємо відсутнє питання

Нехай гіпотеза H2 звучить так:

> Додавання помилок у мітки навчальних даних змінює якість класифікації, причому чутливість різних алгоритмів може відрізнятися.

Уже запропоновано:
- RQ1: Як змінюється якість кожного алгоритму при збільшенні частки помилкових міток?
- RQ2: ______________________________________________
- RQ3: Чи стабільні отримані результати за різних випадкових реалізацій помилок?

**Завдання:** формулюємо RQ2, яке безпосередньо перевіряє порівняльну частину H2. Потім пояснюємо, чому RQ1 недостатньо для перевірки всієї гіпотези.

Відповідь для самоперевірки наведено в розділі H після контрольних питань.

## C6. Від RQ до Research Objectives

**Дослідницьке питання** вказує, що треба дізнатися. **Дослідницьке завдання/ціль** вказує, що треба зробити, щоб це дізнатися.

| Research Question | Відповідна дослідницька дія |
|---|---|
| RQ1. Як змінюється якість кожного алгоритму? | Визначити рівні шуму; виміряти якість на кожному рівні; побудувати криві залежності |
| RQ2. Чи відрізняється деградація якості? | Визначити показник зниження якості; порівняти його між алгоритмами |
| RQ3. Чи стабільні відмінності? | Повторити оцінювання з різними реалізаціями шуму; узагальнити варіативність |

**Відмінність на прикладі:**
- *RQ:* «Чи відрізняється чутливість алгоритмів?»
- *Objective:* «Кількісно порівняти деградацію якості алгоритмів за контрольованих рівнів шуму».

Не підмінюємо дослідницьке завдання прогнозом: «Довести перевагу Random Forest» не є нейтральною ціллю перевірки.

## C7. Попередня логіка перевірки (без детального протоколу)

Для нашого прикладу можна попередньо передбачити:

1. Обрати доступний датасет класифікації з числовими ознаками.
2. Зафіксувати поділ даних та узгоджені правила навчання й оцінювання трьох алгоритмів.
3. Визначити один вид контрольованого шуму та кілька значень його інтенсивності.
4. Оцінити якість кожного алгоритму за однакових умов.
5. Порівняти зміну метрик і повторюваність результатів.
6. Зробити висновок про те, чи підтримують отримані свідчення H1 **в межах вибраного сценарію**.

**Увага до дизайну:** не можна змішувати вплив шуму з неконтрольованими змінами інших умов. У прикладі припускаємо, що шум додається до тестових ознак; варіант із шумом у тренувальних даних — інша постановка, що потребує окремого протоколу.

Точні датасети, метрики, способи формування шуму, повтори та статистичне оцінювання будуть обґрунтовані в наступних спринтах.

## D. Самостійна робота — власна погоджена тема

**У межах цієї вправи нову тему не обираємо.** Використовуємо тему, уже погоджену з керівником. Якщо проблему поки важко уточнити, формулюємо робочу версію та питання для консультації.

**Правила роботи:**
- У відповідях нижче **замінюємо текст у квадратних дужках** власними формулюваннями.
- Не копіюємо формулювання наскрізного прикладу, якщо вони не стосуються обраного дослідження.
- Не подаємо припущення як підтверджені результати.
- Для кожного RQ пояснюємо його роль у перевірці гіпотези.
- Остаточне уточнення проблеми можливе після огляду літератури в Sprint 2.

## E. Sprint Review — checklist готовності

Перед поданням результату керівнику перевіряємо кожен пункт:

- [ ] Тема погоджена з керівником; у спринті її не підмінено іншою.
- [ ] Проблема сформульована як конкретна невизначеність у межах теми.
- [ ] Гіпотеза — перевірюване твердження, для якого можливий результат, що її не підтримує.
- [ ] Сформульовано 2–3 змістовно пов'язані RQ.
- [ ] **Для кожного RQ** пояснено, яку частину гіпотези перевіряє відповідь.
- [ ] **Для кожного RQ** визначено тип потрібного свідчення або порівняння.
- [ ] Дослідницькі цілі сформульовані як дії, а не бажані висновки.
- [ ] Є попередня логіка перевірки та реалістичні межі.
- [ ] Немає вигаданих експериментальних результатів або недоведених тверджень про «найкращий» метод.
- [ ] Питання, які потребують огляду літератури, зафіксовано для Sprint 2.

**Статус спринту:** `READY FOR SUPERVISOR REVIEW`, якщо всі змістовні пункти виконано; інакше `REVISION NEEDED`.

Цей checklist призначений для самоперевірки. Наукову коректність і реалістичність постановки оцінюють студент і науковий керівник.

## F. Підсумковий артефакт: Research Problem Statement

Переносимо остаточні відповіді з розділу D в один стислий документ за структурою:

1. **Research Topic** — погоджена тема.
2. **Problem Context and Research Problem** — контекст та конкретна проблема.
3. **Research Hypothesis** — перевірюване припущення.
4. **Research Questions** — 2–3 питання із поясненням зв'язку з гіпотезою.
5. **Research Objectives** — дії для отримання відповідей.
6. **Preliminary Verification Approach** — загальна логіка перевірки.
7. **Scope, Constraints, Open Questions** — межі, ресурси, питання до Sprint 2.

**Рекомендований обсяг:** 1–2 сторінки змістовного тексту (не враховуючи робочі нотатки в ноутбуці).

**Рекомендоване ім'я файлу:** `Research_Problem_Statement.md`.

Зберігаємо файл у власному fork разом із заповненим ноутбуком. У вихідний (upstream) репозиторій студентські результати не додаються.

Результат Sprint 1 — **робоча постановка**, що стане основою для Research Proposal і літературного обґрунтування у Sprint 2. Вона може бути аргументовано уточнена після знайомства з джерелами.

## G. Підсумок і перехід до Sprint 2

**Тема → Проблема → Гіпотеза → Research Questions → Research Objectives → Попередня перевірка.**

Особливо важливий крок — **гіпотеза → дослідницькі питання**: гіпотезу недостатньо перетворити на питальне речення; потрібно з'ясувати, **які саме свідчення** знадобляться для її перевірки.

**У наступному спринті:** Research Proposal & Literature Review. Ми перевіримо, наскільки проблема обґрунтована у контексті наявних досліджень, і за потреби уточнимо H1 та RQ.

**Контрольні питання:**
1. Чим наукова проблема відрізняється від теми?
2. Чому гіпотеза повинна допускати результат, що її не підтримує?
3. Чому одного питання «Чи правильна H1?» зазвичай недостатньо?
4. Як довести, що кожне RQ необхідне для перевірки H1?
5. Що зміниться, якщо на Sprint 2 буде знайдено дослідження з близькими висновками?

## H. Відповіді до тренувальних вправ — для самоперевірки

### Вправа 1

A — дослідницька дія (побудувати графік); B — гіпотетичне твердження, якщо його ще не перевірено; C — дослідницьке питання; D — передчасний висновок без результатів. Сам по собі тип формулювання не гарантує наукової якості.

### Вправа 2

RQ2: «Чи відрізняється величина зниження якості класифікації між вибраними алгоритмами за однакових часток помилково позначених навчальних прикладів?» Відповідь на RQ1 дає окремі залежності, але ще не формалізує порівняння між алгоритмами.
