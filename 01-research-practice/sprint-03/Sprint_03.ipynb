## A. Problem Situation — чи відповідає експеримент на наше питання?

У Sprint 2 ми уточнили гіпотезу про різну чутливість ML-алгоритмів до шуму у вхідних даних. Можна навчити три моделі, додати шум і порівняти результати. Але чи буде цього достатньо для обґрунтованого висновку?

Необхідно з'ясувати:

- чи працюють алгоритми з однаковими навчальними й тестовими даними;
- що називаємо «шумом» і як вимірюємо його інтенсивність;
- як відрізняємо високу вихідну якість від меншої деградації;
- як враховуємо випадкові реалізації шуму;
- за яких умов результати підтримуватимуть H1 і які висновки будуть недопустимими.

**Наскрізна проблема спринту:** навіть правильно сформульована гіпотеза не буде перевірена, якщо експеримент не відокремлює досліджуваний фактор від сторонніх впливів.

## B1. Essential Theory — від Research Question до спостереження

Research Proposal пояснює, *чому* дослідження має сенс. Experimental Protocol пояснює, *як* отримати дані для перевірки гіпотези.

Переходимо за логікою:

**H1 → Research Question → необхідне порівняння → фактори та контроль → вимірювання → інтерпретація.**

**Операціоналізація** — це визначення способу виміряти поняття. Наприклад, «стійкість алгоритму» можна операціоналізувати як зміну macro-F1 за зростання інтенсивності контрольованого шуму.

| Компонент | Що визначаємо |
|---|---|
| Research Question | Яку властивість або залежність потрібно з'ясувати? |
| Порівняння | Які умови повинні відрізнятися, а які залишатися однаковими? |
| Вимірювання | Яка метрика та за яких умов дає відповідь? |
| Інтерпретація | Що означатимуть підтримка, непідтримка чи неоднозначність результатів? |

Окреме значення метрики за одного запуску зазвичай не показує, *як* модель реагує на зростання шуму.

## B2. Essential Theory — змінні та контроль експерименту

У контрольованому експерименті розрізняємо:

- **Independent variables (фактори)** — що змінюємо: алгоритм класифікації та інтенсивність тестового шуму.
- **Dependent variable** — що вимірюємо: macro-F1 та зменшення якості відносно стану без шуму.
- **Controlled variables** — що утримуємо однаковим: джерело даних, тестове розбиття, механізм шуму, процедура оцінювання.
- **Сторонні чинники** — що може спотворити порівняння: різні тестові приклади, витік даних, нерівноцінне налаштування моделей.

Порівняння різних алгоритмів не вимагає однакових гіперпараметрів — це часто неможливо. Натомість визначаємо **порівнювану процедуру налаштування** або обґрунтовані фіксовані конфігурації.

### Повторні вимірювання та можливе розширення експерименту

Оскільки шум генерується випадково, навіть за однакової інтенсивності можемо отримати дещо різні значення метрики. Тому для кожного рівня шуму повторюємо вимірювання, щоразу генеруючи нову реалізацію шуму. При цьому навчена модель і тестовий набір залишаються незмінними — змінюються лише випадкові спотворення його ознак.

Таким чином, оцінюємо варіативність результатів, зумовлену випадковими спотвореннями вхідних даних. Однак питання, чи збережуться виявлені закономірності на інших вибірках, залишається відкритим, оскільки повторні розбиття одного набору даних не замінюють перевірку на незалежних датасетах.

**Можливим напрямом подальшого розвитку дослідження** є розширення експериментального протоколу: використання інших розбиттів даних або незалежних наборів даних із повторним навчанням моделей. Це дозволить додатково оцінити залежність результатів від вибору даних та перевірити узагальнюваність отриманих висновків. 

## B3. Essential Theory — baseline та чесне порівняння

**Baseline Model** — заздалегідь визначений опорний алгоритм, відносно якого можна оцінювати інші методи. У нашому прикладі це — **Logistic Regression**. Її обрано як відносно просту лінійну модель для зіставлення з ансамблевими моделями.

Окремо для **кожної** досліджуваної моделі визначаємо Clean Reference Performance. 

**Clean Reference Performance** — базовий рівень якості класифікації, отриманий під час тестування навченої моделі на даних без штучного шуму.

Цей показник використовуємо для визначення зменшення якості класифікації кожного алгоритму за різних рівнів інтенсивності шуму. 

Отже, маємо два **різні види порівнянь**:

1. **Між алгоритмами:** Logistic Regression, Random Forest та Gradient Boosting.
2. **У межах кожного алгоритму:** його якість за наявності шуму порівняно з власною якістю без шуму.

Для коректного порівняння алгоритмів використовуємо однаковий тестовий набір даних (Test Set), метрики оцінювання та правила генерації шуму.

Навчання моделей і підбір їхніх гіперпараметрів виконуємо на навчальних і валідаційних даних (Training Set / Validation Set або Cross-Validation). Після завершення налаштування фіксуємо конфігурації моделей.

**Test Set використовуємо виключно для остаточного оцінювання якості та її зміни під впливом шуму.** Результати тестування не повинні впливати на вибір гіперпараметрів або подальше налаштування моделей, оскільки це порушує незалежність оцінювання.

## B4. Essential Theory — метрики, повторення й аналіз невизначеності

Метрика повинна вимірювати властивість, визначену в Research Question. Для класифікації можуть бути доречними Accuracy, Precision, Recall, F1-score або ROC AUC — залежно від постановки задачі та класового балансу.

У наскрізному прикладі обираємо **macro-F1**, який обчислює F1 окремо для кожного класу та усереднює їх із рівними вагами. Це дає змогу не зводити оцінку лише до найчисленнішого класу.

Позначимо якість моделі `m` за інтенсивності тестового шуму `σ` через `Q(m, σ)`. Використовуємо **два показники**:

- **Якість класифікації:** `Q(m, σ)`
- **Зменшення якості:** `D(m, σ) = Q(m, 0) − Q(m, σ)`,
  де $Q(m,0)$ — Clean Reference Performance.

Якщо `D(m, σ) > 0`, якість погіршилася; якщо `D(m, σ) < 0`, оцінка якості підвищилася. Значення `D` саме по собі не доводить перевагу алгоритму: враховуємо вихідну якість `Q(m, 0)` та варіативність результатів.

Оскільки шум випадковий, за кожного додатного `σ` повторюємо його генерацію з **наперед визначеними seed**. Для тих самих даних і test split різні реалізації шуму відображають варіативність *внесених спотворень*, але не узагальнюють результати на всі можливі дані.

**Важливо:** Метрику, рівні шуму, seed, спосіб узагальнення та принципи порівняння фіксуємо **до виконання основних вимірювань**, а не обираємо за вигідними результатами.

## B5. Essential Theory — Experimental Protocol та відтворюваність

**Experimental Protocol** — документ, який однозначно визначає заплановану процедуру отримання й аналізу результатів.

Він описує:

1. Дані, їхнє походження, версію та обмеження використання.
2. Поділ на train/validation/test і спосіб уникнення data leakage.
3. Моделі, baseline, препроцесинг і правила tuning.
4. Фактори, діапазони значень та контрольовані умови.
5. Метрики, повторення, seed та аналіз невизначеності.
6. Зв'язок результатів із Research Questions і H1.
7. Програмне середовище, конфігурації та спосіб відтворення.
8. Обробку помилок запуску та документування відхилень від плану.

**Відтворюваність результатів** потребує не лише фіксації початкових значень генераторів псевдовипадкових чисел (random seeds), а й можливості з'ясувати, які дані, версії бібліотек і параметри використовувалися. Результат Sprint 3 — план; код і фактичні вимірювання з'являться в наступних спринтах.

## B6. Essential Theory — загрози валідності та межі майбутніх висновків

Важливо розрізняти дві речі:

- **Загрози валідності** — фактори, які можуть спотворити результати або їхню інтерпретацію.
- **Межі застосовності висновків** — умови, поза якими отримані результати не можна автоматично узагальнювати.

### Загрози валідності

| Загроза | Можливий наслідок | Запобіжний захід |
|---|---|---|
| Data leakage | Тестові дані впливають на підготовку моделі | Розділяємо дані до препроцесингу; параметри визначаємо лише на train |
| Відмінності в умовах тестування | Відмінність пояснюється різними тестовими прикладами | Спільний test split і ті самі реалізації шуму для всіх алгоритмів |
| Некоректна метрика | Погіршення для окремого класу приховано загальною Accuracy | Обґрунтовуємо macro-F1 або іншу придатну метрику |
| Випадковість шуму | Результат однієї реалізації шуму може не відображати характерної поведінки моделі за заданих умов | Повторні вимірювання з різними seed, оцінювання середнього значення та розкиду результатів |
| Post-hoc selection | Залишаємо тільки вигідні рівні шуму | Фіксуємо матрицю та аналіз до експерименту |

### Межі застосовності майбутніх висновків

Під час планування експерименту необхідно заздалегідь встановити, **на які об'єкти, умови та властивості поширюватимуться отримані результати**.

Для цього послідовно визначаємо:

1. **Об'єкти дослідження.** Які алгоритми, моделі або системи порівнюємо?
2. **Дані.** На яких наборах даних проводимо експеримент і які їхні характеристики?
3. **Експериментальний вплив.** Який саме фактор змінюємо і в яких межах?
4. **Показники оцінювання.** За якими метриками вимірюємо результати?
5. **Умови проведення.** Які параметри та процедури залишаються незмінними?
6. **Джерела варіативності.** Які випадкові фактори враховуємо під час повторних вимірювань?

Відповіді на ці питання дозволяють сформулювати **область застосовності майбутніх висновків (Scope of Conclusions)**.

Наприклад, навіть якщо ми повністю усунули помилки в організації експерименту, це ще не означає, що його результати залишатимуться справедливими для інших алгоритмів, даних або типів шуму.

Отже, після визначення експериментальних умов формулюємо, **які саме твердження зможемо обґрунтувати на основі результатів, а які потребуватимуть додаткових досліджень**.

## C1. Guided Research Example — від Research Proposal до запланованих спостережень

**Тема:** експериментальне дослідження стійкості алгоритмів ML до шуму вхідних ознак.

**Гіпотеза H1:** за спільних умов навчання й тестування Logistic Regression, Random Forest і Gradient Boosting демонструватимуть різні профілі деградації якості зі збільшенням **шуму в тестових числових ознаках**.

Усі моделі навчаються **на даних без штучного шуму**. Після навчання до тестових ознак додаємо адитивний гаусів шум. **Під час шумових тестувань моделі не перенавчаємо.**

| Research Question | Які результати потрібні | Яке порівняння виконуємо |
|---|---|---|
| RQ1. Як змінюється якість кожного алгоритму зі зростанням шуму? | `Q(m, σ)` для кожного рівня | Криві якості за `σ` |
| RQ2. Чи відрізняються профілі деградації між алгоритмами? | `D(m, σ)` | Криві деградації між моделями |
| RQ3. Наскільки стійкі спостереження до випадкових реалізацій шуму? | Результати за різних seed | Розкид оцінок для кожного `m, σ` |

Для RQ1 потрібна **серія вимірювань**, для RQ2 — **зіставлення змін**, для RQ3 — **повторення**. Саме цю структуру реалізує матриця експериментів у С4.

## C2. Guided Research Example — операціоналізуємо шум і стійкість

У нашому дослідженні перевіряємо, як змінюється якість класифікації, коли вхідні числові ознаки **тестових даних** стають менш точними.

### 1. Підготовка даних

Припустимо, що числові ознаки описують температуру, тиск і вібрацію обладнання. Вони мають різні одиниці й масштаби. Однакове за абсолютною величиною спотворення могло б мати для них різний відносний вплив. Тому перед внесенням шуму виконуємо **стандартизацію**.

Для кожної ознаки:

$$
z = \frac{x-\mu}{s}
$$

де $x$ — початкове значення, $\mu$ — середнє, $s$ — стандартне відхилення, $z$ — стандартизована ознака.

**Порядок дій важливий.** Спочатку поділяємо дані на **Training Set** (навчання) і **Test Set** (незалежна перевірка). Параметри стандартизації $\mu$ та $s$ визначаємо **лише за Training Set**. Потім застосовуємо ті самі параметри до Test Set. Під час cross-validation параметри визначаються окремо всередині кожної навчальної складки. Це не дозволяє тестовим даним впливати на підготовку моделі (*data leakage*).

### 2. Модель шуму

До **стандартизованих тестових** ознак додаємо випадкове відхилення:

$$
x_{\mathrm{noisy}} = x_{\mathrm{clean}} + \varepsilon
$$

Тут $x_{\mathrm{clean}}$ — значення ознаки без штучного шуму, $x_{\mathrm{noisy}}$ — після внесення шуму, $\varepsilon$ — випадкове спотворення.

Обираємо нормальний (гаусів) розподіл:

$$
\varepsilon \sim \mathcal{N}(0,\sigma^2)
$$

Це **позначення розподілу ймовірностей**, а не виклик функції: $0$ означає відсутність систематичного зміщення шуму; $\sigma^2$ — дисперсію; $\sigma$ — стандартне відхилення, що визначає характерну величину спотворень.

Рівні шуму:

$$
\sigma \in \{0,\ 0.1,\ 0.2,\ 0.4,\ 0.6\}
$$

За $\sigma=0$ штучний шум відсутній. Більше $\sigma$ зазвичай означає сильніше спотворення. Оскільки шум вносимо **після стандартизації**, ці рівні задано в масштабі стандартизованих ознак, а не в початкових фізичних одиницях. Для кожного тестового об'єкта й кожної ознаки генеруємо незалежне відхилення.

### 3. Оцінювання стійкості

Моделі навчаємо **один раз** на підготовлених навчальних даних без штучного шуму (після остаточного вибору конфігурацій). На зафіксованому Test Set обчислюємо `Q(m, 0)` та `Q(m, σ)` за додатних рівнів шуму; за кожного додатного рівня повторюємо генерацію шуму з різними seed.

Зменшення якості визначаємо так само, як у B4:

$$
D(m,\sigma) = Q(m,0) - Q(m,\sigma)
$$

Отже, **шум є контрольованим впливом**, `macro-F1` — вимірюваною якістю, `D(m, σ)` — її зменшенням відносно стану без шуму. Ми досліджуємо стійкість **навченої моделі під час застосування**, а не вплив шуму на її навчання.

## C3. Guided Research Example — baseline, поділ даних і налаштування моделей

У C2 ми визначили механізм шуму. Тепер забезпечимо коректне порівняння алгоритмів.

### 1. Моделі

- **Logistic Regression** — наша **Baseline Model**, опорний лінійний класифікатор.
- **Random Forest** — ансамбль дерев рішень.
- **Gradient Boosting** — ансамбль із послідовним навчанням моделей.

Baseline — це орієнтир для міжмодельного порівняння, а не наперед визначений «найгірший алгоритм».

### 2. Поділ даних

Перед обчисленням статистик стандартизації резервуємо **Test Set** (наприклад, 20%). Решту даних використовуємо для навчання та добору гіперпараметрів через validation або cross-validation. Для класифікації за можливості зберігаємо співвідношення класів (*stratified split*).

Якщо дані мають часові або групові залежності, обираємо відповідний time-aware або group-aware поділ замість випадкового перемішування.

### 3. Підготовка й налаштування моделей

Після поділу даних на Training Set і Test Set необхідно підготувати дані для навчання алгоритмів та визначити конфігурації моделей.

**Підготовка даних (Preprocessing).** У нашому експерименті використовуємо стандартизацію числових ознак, розглянуту в C2. Середнє значення та стандартне відхилення кожної ознаки обчислюємо за навчальними даними, після чого застосовуємо отримане перетворення до тестових даних.

Для організації цієї процедури можна використати `Pipeline` із бібліотеки scikit-learn, який об'єднує попередню обробку даних і модель класифікації в єдину послідовність операцій.

Це особливо важливо під час Cross-Validation: параметри стандартизації повинні визначатися лише за тією частиною даних, яка використовується для навчання на кожному кроці валідації. Такий підхід допомагає запобігти **Data Leakage**.

**Налаштування моделей (Model Configuration).** Якість класифікації та стійкість алгоритмів до шуму можуть залежати від значень їхніх гіперпараметрів. Тому перед основним експериментом визначаємо обґрунтовані конфігурації моделей шляхом оптимізації гіперпараметрів на навчальних даних (Hyperparameter Tuning).

Для Logistic Regression, Random Forest і Gradient Boosting задаємо відповідні простори пошуку гіперпараметрів та застосовуємо Cross-Validation. Як критерій вибору конфігурації використовуємо macro-F1 на даних без штучного шуму.

Для коректного порівняння застосовуємо однакову схему валідації та заздалегідь визначені, зіставні за обчислювальними витратами процедури пошуку. Це не означає використання однакових гіперпараметрів, оскільки алгоритми мають різну структуру.

Після завершення оптимізації фіксуємо вибрані гіперпараметри та навчаємо остаточні моделі на Training Set без штучного шуму.

У подальших експериментах конфігурації та навчені моделі залишаються незмінними. Змінюємо лише інтенсивність і випадкові реалізації шуму в Test Set.

Test Set не використовуємо для оптимізації гіперпараметрів. Обрані конфігурації, простори пошуку, критерій оптимізації та процедуру Cross-Validation документуємо в Experimental Protocol.

### 4. Початкова якість без шуму

Для кожної моделі вимірюємо власну **Clean Reference Performance** — `Q(m, 0)` на Test Set без штучного шуму. Це — базовий рівень якості для розрахунку `D(m, σ)`, а не ще одна baseline-модель.

Отже, між собою порівнюємо три моделі, а всередині кожної моделі — якість за різних рівнів тестового шуму. В усіх умовах використовуємо той самий Test Set і ту саму метрику.

## C4. Guided Research Example — матриця експериментів (Experimental Matrix)

У C2 визначили спосіб внесення шуму, у C3 — моделі та правила підготовки даних. Тепер об'єднуємо їх у **матрицю експериментальних умов**.

### 1. Що змінюємо, а що фіксуємо?

**Змінюємо:** алгоритм `m` та інтенсивність **тестового** шуму `σ`.

**Фіксуємо:** датасет і split, процедуру стандартизації, навчену конфігурацію кожної моделі, метрику, механізм шуму, Test Set. За однакового `σ` і seed застосовуємо **одну й ту саму шумову реалізацію до тестових ознак для всіх трьох моделей**, щоб порівняння було парним.

**Рівні шуму:** `σ = 0; 0.1; 0.2; 0.4; 0.6`.

### 2. Матриця умов

| Алгоритм `m` | `σ = 0` | `σ = 0.1` | `σ = 0.2` | `σ = 0.4` | `σ = 0.6` |
|---|---|---|---|---|---|
| Logistic Regression | `Q(LR,0)` | `Q(LR,0.1)` | `Q(LR,0.2)` | `Q(LR,0.4)` | `Q(LR,0.6)` |
| Random Forest | `Q(RF,0)` | `Q(RF,0.1)` | `Q(RF,0.2)` | `Q(RF,0.4)` | `Q(RF,0.6)` |
| Gradient Boosting | `Q(GB,0)` | `Q(GB,0.1)` | `Q(GB,0.2)` | `Q(GB,0.4)` | `Q(GB,0.6)` |

**Маємо 15 унікальних умов:** 3 алгоритми × 5 рівнів шуму. Комірки — **показники, які плануємо виміряти**, а не вигадані значення якості.

### 3. Повторення

Для кожного додатного рівня шуму плануємо **5 реалізацій** з наперед визначеними seed, наприклад `11, 23, 37, 41, 53`. За `σ=0` шум відсутній, тому для фіксованої навченої моделі потрібна одна чиста оцінка.

Отже, для цього плану отримуємо:

- **3 чисті оцінки** `Q(m,0)`;
- **3 × 4 × 5 = 60 шумових оцінок** `Q(m,σ)`;
- **разом 63 обчислення метрики** на зафіксованому Test Set.

Це **не 63 незалежні вибірки**: тестові об'єкти залишаються тими самими, а моделі не перенавчаються. П'ять повторень оцінюють варіативність внесеного шуму за фіксованих умов. Кількість повторень слід обґрунтувати щодо точності оцінок і доступних ресурсів.

### 4. Що обчислюємо з матриці?

- Для **RQ1** показуємо залежності `Q(m,σ)` від `σ`.
- Для **RQ2** обчислюємо `D(m,σ) = Q(m,0) − Q(m,σ)` і порівнюємо профілі деградації між алгоритмами.
- Для **RQ3** показуємо варіативність `Q` і `D` між шумовими реалізаціями для кожного `m,σ`.

**Зауваження**: Статистичні інтервали та висновки про міжмодельні відмінності потрібно розраховувати методами, що враховують **парність** порівнянь і залежність повторних вимірювань на одному тестовому наборі. Самої кількості запусків недостатньо для висновку про статистичну значущість.

### 5. Межі застосовності майбутніх висновків

Використовуючи визначені експериментальні умови, встановимо область застосовності результатів нашого дослідження.

| Аспект | Межі нашого експерименту |
|---|---|
| Алгоритми | Logistic Regression, Random Forest, Gradient Boosting |
| Дані | Обраний набір даних із числовими ознаками |
| Навчання | Моделі навчаються без штучного шуму |
| Тип шуму | Адитивний гаусівський шум у тестових даних |
| Інтенсивність шуму | $\sigma \in \{0,\ 0.1,\ 0.2,\ 0.4,\ 0.6\}$ |
| Метрика | Macro-F1 |
| Повторення | Декілька реалізацій шуму за фіксованих Training Set і Test Set |

Визначені умови дозволяють дослідити чутливість обраних алгоритмів до гаусівського шуму на конкретному наборі даних.

Водночас отримані результати **не дозволятимуть автоматично поширювати висновки** на інші алгоритми, незалежні набори даних, інші типи шуму або спотворення навчальних даних.

Після виконання експерименту зможемо сформулювати висновок такого типу:

> За визначених експериментальних умов встановлено [опис характеру відмінностей] між профілями деградації якості досліджуваних алгоритмів.

Конкретний зміст висновку визначатиметься лише отриманими результатами.

Таким чином, ми заздалегідь окреслюємо область, у межах якої перевірятимемо H1, і не приписуємо результатам значення ширше, ніж дозволяє експеримент.

## C5. Guided Research Example — інтерпретація результату

У C4 спланували вимірювання. Тепер визначимо **до отримання результатів**, як вони будуть пов'язані з перевіркою H1.

**H1:** профілі деградації macro-F1 зі збільшенням тестового шуму відрізнятимуться між Logistic Regression, Random Forest і Gradient Boosting за спільних умов експерименту.

Для кожної моделі аналізуємо **абсолютну якість** `Q(m,σ)` і **зменшення якості** `D(m,σ) = Q(m,0) − Q(m,σ)`.

### Можливі сценарії інтерпретації щодо H1

**1. Результати підтримують H1.** Спостерігаємо змістовні відмінності профілів `D(m,σ)` між алгоритмами; відмінності достатньо стійкі до реалізацій шуму відповідно до наперед визначеного способу оцінювання невизначеності. Висновок стосується лише перевірених даних, моделей і шумового механізму.

**2. Результати не надають підтримки H1.** Спостережувані профілі подібні або встановлені відмінності не відповідають наперед визначеним критеріям. Це **не тотожне доказу однакової стійкості**: для твердження про практичну еквівалентність потрібні відповідна межа еквівалентності та достатня точність оцінок.

**3. Результати неоднозначні.** Розкид або невизначеність не дозволяє відрізнити можливу міжмодельну різницю від випадкових коливань. Потрібно повідомити про невизначеність, а не оголошувати H1 підтвердженою чи спростованою.

### Повертаємося до Research Questions

- **RQ1:** що показують криві `Q(m,σ)`?
- **RQ2:** як відрізняються криві `D(m,σ)`?
- **RQ3:** наскільки стійкі результати до повторних реалізацій шуму?

**Важливо:** навіть якщо два алгоритми мають однакову якість за певного `σ`, їхня деградація може відрізнятися через різні `Q(m,0)`. Тому інтерпретуємо `Q` і `D` разом. Умови і правила оцінювання H1 погоджуємо **до** запуску основного експерименту.

## Тренувальна вправа 1 — чи відповідає експеримент на RQ?

**Ситуація:** RQ: «Чи відрізняється стійкість двох класифікаторів до шуму тестових ознак?» Модель A навчено та перевірено на одному наборі даних, модель B — на іншому. Для обох додано шум різної інтенсивності й порівняно Accuracy.

1. Визначаємо щонайменше **три причини**, через які не можна приписати спостережувану відмінність лише алгоритму.
2. Пропонуємо мінімальну зміну дизайну для контрольованого порівняння.
3. Формулюємо допустиму область майбутнього висновку.

Відповіді для самоперевірки наведено в розділі G після контрольних питань.

## Тренувальна вправа 2 — метрика та деградація

**Ситуація:** Для моделі A маємо `Q(A,0)=0.92`, `Q(A,0.2)=0.83`; для моделі B — `Q(B,0)=0.86`, `Q(B,0.2)=0.81`.

1. Обчислюємо `D(A,0.2)` та `D(B,0.2)`.
2. Порівнюємо абсолютну якість моделей за `σ=0.2` та їхню деградацію.
3. Пояснюємо, чому одиничне вимірювання ще не доводить надійної відмінності профілів стійкості.

Відповіді для самоперевірки наведено в розділі G після контрольних питань.

## Тренувальна вправа 3 — план повторень

**Ситуація:** Після навчання трьох фіксованих моделей плануємо п'ять рівнів тестового шуму: `0; 0.1; 0.2; 0.4; 0.6`. Для кожного додатного рівня генеруємо 5 незалежних реалізацій шуму на тому самому Test Set.

1. Визначаємо кількість унікальних умов.
2. Обчислюємо кількість фактичних оцінювань метрики, якщо для `σ=0` її обчислюємо один раз на модель.
3. Пояснюємо, чи можна назвати всі оцінювання незалежними вибірками.
4. Визначаємо, чи потрібно перенавчати моделі для перевірки нашої H1.

Відповіді для самоперевірки наведено в розділі G після контрольних питань.

## G. Solutions to Training Exercises — відповіді для самоперевірки

### Вправа 1

Різні датасети можуть мати різні розподіли класів, складність, масштаб і склад ознак. Додатково заважають різні split, процедури tuning та реалізації шуму. Для виправлення порівнюємо обидві моделі за **спільного Test Set**, узгоджених налаштувань і **тих самих реалізацій шуму** на кожному рівні `σ`. Допустимий висновок стосується лише досліджених даних, моделей та шумового механізму.

### Вправа 2

`D(A,0.2) = 0.92 − 0.83 = 0.09`; `D(B,0.2) = 0.86 − 0.81 = 0.05`. За `σ=0.2` абсолютна якість A вища (`0.83 > 0.81`), але в B спостерігаємо **меншу деградацію** (`0.05 < 0.09`). Одного вимірювання недостатньо для висновку про стійкий характер відмінності між профілями; потрібні повторення та оцінка невизначеності.

### Вправа 3

Умов: `3 × 5 = 15`. Фактичних оцінювань: `3` чисті оцінки плюс `3 × 4 × 5 = 60` шумових, разом **63**. Вони не є незалежними вибірками: використовуємо той самий Test Set, а моделі фіксовані. Для H1 моделі **не перенавчаємо** за кожного рівня тестового шуму; повторюємо тільки генерацію спотворень та оцінювання.
